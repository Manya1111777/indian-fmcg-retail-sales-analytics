```python
#roduct category generates the highest total revenue? Also find the total units sold and total margin for each category, and rank the categories from highest to lowest revenue.
df.groupby("Category")[["Revenue", "Margin", 'Units']].sum().round(2).sort_values("Revenue", ascending= False)
```

```python
df["Customer_Gender"] = df["Customer_Gender"].fillna("Unknown")
```

```python
print(df["Customer_Gender"].isnull().sum())
print(df["Customer_Gender"].value_counts(dropna=False))
```

```python
df.groupby("Customer_Gender").agg(Transactions= ("Invoice_ID", "count"), Total_Revenue = ("Revenue", "sum")).round(2)
```

```python
df.groupby("Category").agg(Total_Margin= ("Margin", "sum"), Average_Margin= ("Margin", "mean")).round(2).sort_values("Average_Margin", ascending = False)
```

```python
plt.figure(figsize = (10,6))
plt.title("Revenue Transaction Variance")
sns.boxplot(data = df, x = "Revenue")
plt.show()
```

```python
plt.figure(figsize= (10,6))
plt.title("Correlation Heatmap")
correlation = df[["Units", "Revenue", "Cost", "Margin","Stock_On_Hand","Reorder_Level", "Lead_Time_Days"]].corr()
sns.heatmap(correlation, annot = True, cmap = "coolwarm", )
fmt = ".2f"
plt.show()
```

```python
df.groupby("Loyalty_Flag").agg(Transactions=("Invoice_ID", "count"), Total_Revenue=("Revenue", "sum"), Avg_Transaction = ("Revenue", "mean")).round(2)
```

```python
category_revenue = (df.groupby("Category")["Revenue"].sum().sort_values(ascending=False))
category_revenue
```

```python
plt.figure(figsize=(10, 6))
sns.barplot( x=category_revenue.index,y=category_revenue.values)
plt.title("Total Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Total Revenue")
plt.show()
```
